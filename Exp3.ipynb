```python
import pandas as pd

# --------------------------------------------------
# 1. Create a sample dataset
# --------------------------------------------------

data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Bob'],
    'age': [25, 32, 28, 45, 22, 35, 32],
    'salary': [40000, 55000, 48000, 75000, 35000, 60000, 55000],
    'category': ['A', 'B', 'A', 'B', 'A', 'B', 'B'],
    'score': [75, 90, 65, 85, 55, 95, 90]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)


# --------------------------------------------------
# 2. Display first 5 rows
# --------------------------------------------------

print("\nFirst 5 rows:")
print(df.head())


# --------------------------------------------------
# 3. Display last 5 rows
# --------------------------------------------------

print("\nLast 5 rows:")
print(df.tail())


# --------------------------------------------------
# 4. Check data types and general information
# --------------------------------------------------

print("\nDataFrame Information:")
df.info()


# --------------------------------------------------
# 5. Summary statistics
# --------------------------------------------------

print("\nSummary Statistics:")
print(df.describe())


# --------------------------------------------------
# 6. Handle missing values
# --------------------------------------------------

df['salary'] = df['salary'].fillna(df['salary'].mean())
df['score'] = df['score'].fillna(df['score'].mean())

print("\nAfter handling missing values:")
print(df)


# --------------------------------------------------
# 7. Create a new column
# --------------------------------------------------

df['double_score'] = df['score'] * 2

print("\nNew column - double_score:")
print(df)


# --------------------------------------------------
# 8. Create a Series and perform operations
# --------------------------------------------------

series = df['score']

print("\nOriginal Series:")
print(series)

print("\nSeries + 10:")
print(series + 10)


# --------------------------------------------------
# 9. Filter rows based on conditions
# --------------------------------------------------

filtered_df = df[
    (df['score'] > 50) &
    (df['salary'] < 70000)
]

print("\nFiltered DataFrame:")
print(filtered_df)


# --------------------------------------------------
# 10. Grouping and aggregation
# --------------------------------------------------

grouped = df.groupby('category')['score'].mean()

print("\nGrouped Mean Score:")
print(grouped)


# --------------------------------------------------
# 11. Sorting
# --------------------------------------------------

df_sorted = df.sort_values(
    by='score',
    ascending=False
)

print("\nSorted DataFrame:")
print(df_sorted)


# --------------------------------------------------
# 12. Boolean masking
# --------------------------------------------------

median_score = df['score'].median()

masked_df = df[
    df['score'] > median_score
]

print("\nMedian Score:", median_score)

print("\nMasked DataFrame:")
print(masked_df)


# --------------------------------------------------
# 13. Remove duplicates
# --------------------------------------------------

df.drop_duplicates(inplace=True)

print("\nAfter removing duplicates:")
print(df)


# --------------------------------------------------
# 14. Create a DataFrame with selected columns
# --------------------------------------------------

subset_df = df[['name', 'score']]

print("\nSelected Columns:")
print(subset_df)


# --------------------------------------------------
# 15. Save DataFrame to CSV
# --------------------------------------------------

subset_df.to_csv(
    'filtered_data.csv',
    index=False
)

print("\nfiltered_data.csv has been created successfully!")


# --------------------------------------------------
# 16. Summary calculations
# --------------------------------------------------

print("\nFinal Statistics:")

print("Total Score:", df['score'].sum())

print("Mean Score:", df['score'].mean())

print("Standard Deviation:", df['score'].std())
```
